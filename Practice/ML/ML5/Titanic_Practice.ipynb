## 2.Fixing the placeholder
in this case its "?" it can be anything
keep the original df as it is and create a copy and work on it 

````python
df_clean = df_original.copy()
````

````python
print(f"Cells containing '?' before: {(df_clean == '?').sum().sum()}")
df_clean = df_clean.replace('?',np.nan)
print(f"Cells containing '?' after: {(df_clean == '?').sum().sum()}")
````

````python
print("=== ORIGINAL ===")
df_original.info()

print("\n=== CLEAN ===")
df_clean.info()
````

````python
# Shows non-null count for both side by side
#df.count() counts the non-null values per column — it excludes NaN but counts everything else, including ?.
pd.DataFrame({
    'original': df_original.count(),
    'clean': df_clean.count()
})
````

# dtype conversion
 identified age,fare and body as object/string it should be number.

````python
'''
It converts to float64 by default when the column contains NaN, and int64 when there are no NaNs.

In your case all three columns (age, fare, body) have NaNs from the ? replacement, so they will all become float64.

The reason is what concept 2 mentions: standard integers cannot hold NaN in pandas. So any numeric column with missing values automatically becomes float64. That’s expected, not an error.
'''

numeric_columns = ['age','fare','body']

for column in numeric_columns:
    df_clean[column] = pd.to_numeric(df_clean[column], errors='coerce')  #Without errors='coerce', pd.to_numeric() uses errors='raise' by default — it stops and throws an error the moment it hits a value it can’t convert.


df_clean.info()
````

## Measuring Missingness

### What each line does

**Line 1 — Count missing values per column**
```python
missing_count = df_clean.isna().sum().sort_values(ascending=False)
```
- `df_clean.isna()` — marks every cell `True` if missing, `False` if not   [df_clean.isna().sum() gives you one number per column: the total missing values in that]
- `.sum()` — counts the `True` values per column (True=1, False=0)
- `.sort_values(ascending=False)` — sorts columns from most missing to least
- Result: a Series with missing count per column, worst first

**Line 2 — Convert counts to percentages**
```python
missing_percentage = (missing_count / len(df_clean) * 100).round(2)
```
- `missing_count / len(df_clean)` — divides each count by total rows (1,309)
- `* 100` — converts to percentage
- `.round(2)` — keeps 2 decimal places
- Result: same Series but as percentages

**Line 3 — Build the summary table**
```python
missing_summary = pd.DataFrame({
    'missing_count': missing_count,
    'missing_percentage': missing_percentage
})
```
- `pd.DataFrame({...})` — builds a table from both Series
- The keys become column headers
- Both Series share the same index (column names), so each row lines up correctly
- Result: one table, one row per feature

## 6 domain check 

### Check 1 — Survival rate of passengers who had a boat:

````python
has_boat = df_clean['boat'].notna()
survival_rate_with_boat = df_clean.loc[has_boat, 'survived'].mean() * 100
survival_rate_without_boat = df_clean.loc[~has_boat, 'survived'].mean() * 100

print(f'Survival rate WITH a boat:    {survival_rate_with_boat:.2f}%')
print(f'Survival rate WITHOUT a boat: {survival_rate_without_boat:.2f}%')
````

98.15% of passengers who had a boat value survived.

A lifeboat assignment is an almost perfect signal of survival.
Only 9 people who had a boat did not survive — those are suspicious rows worth investigating.

2.79% of passengers without a boat survived.

Almost everyone without a boat died.
The small 2.79% are likely people who survived by other means — pulled from the water, for example.

The domain rule this confirms:
A boat value → almost certainly survived. No boat value → almost certainly did not.

### 2. contradiction check — passengers who had a boat but did not survive

````python
died_with_boat = df_clean[has_boat & (df_clean['survived'] == 0)]

print(f'Passengers with a boat who did not survive: {len(died_with_boat)}')
died_with_boat[['name', 'pclass', 'survived', 'boat']]
````

A few observations:

The Lindell couple (rows 968, 969) — husband and wife, both on boat A, both marked as not survived. Likely a data entry error — historically, Mrs. Lindell survived but Mr. Lindell did not. The wife’s record is probably wrong.

Boat A appears 4 times — Beattie, Keefe, Lindell ×2 all show boat A. Boat A was a collapsible lifeboat that had a complicated story — some people on it didn’t make it. So this may not be a data error but a real edge case.

What to do with these rows:

They are flagged, not fixed at this stage.
In a real project you’d investigate each one against historical records.
For this exercise, we note them and move on — the domain check has done its job.

## 6. categorical features

## Categorical Features

### Standard checks for each categorical column

| Step | Code | Purpose |
|---|---|---|
| 1. Count unique categories | `df_clean['col'].nunique()` | How many distinct values exist |
| 2. List the categories | `df_clean['col'].unique()` | What those values actually are |
| 3. Frequency of each category | `df_clean['col'].value_counts()` | How common each value is |
| 4. Frequency including missing | `df_clean['col'].value_counts(dropna=False)` | Spots unexpected values and NaN in one call |
| 5. Percentage per category | `df_clean['col'].value_counts(normalize=True) * 100` | Relative size of each category |

> Steps 3 and 4 together are the most useful — run them on every categorical column.

---

### Three questions to ask per column

**1. Are the values valid?**
- No typos, no inconsistent casing, no unexpected categories
- If found → fix in cleaning (e.g. `df_clean['sex'].str.lower()`)

**2. Are there missing values?**
- If yes → fill in imputation stage
- If no → nothing to do

**3. Is the column usable for a model?**
- Categorical text columns need encoding (text → numbers)
- But encoding happens in **preprocessing**, not here

---

### Fix now vs fix later

| Situation | Action | Stage |
|---|---|---|
| Inconsistent labels (`'Male'`, `'male'`, `'MALE'`) | Fix | Cleaning |
| Typos (`'Southhampton'` instead of `'S'`) | Fix | Cleaning |
| Text that needs converting to numbers | Encode | Preprocessing |
| Rare categories to merge | Combine | Preprocessing |
| Dummy variables / one-hot encoding | Create | Preprocessing |

> **Rule: quality problems → fix in cleaning. Form changes → leave for preprocessing.**

---

### Why not encode here?

- Encoding loses the original labels (`male/female` becomes `0/1`) — harder to read and debug
- Preprocessing can be changed without touching cleaning
- Different encoding strategies can be tried without redoing cleaning

---

## 7. Imputation
Replacing the missing values using a stated rules


````python
# Age — fill with mode
age_mode = df_drop['age'].mode().iloc[0]   # .iloc[0]  - since the mode can return many value i am asking to give me first one 
print(f'Age mode: {age_mode}')
df_drop['age'] = df_drop['age'].fillna(age_mode)

# Embarked — fill with mode
embarked_mode = df_drop['embarked'].mode().iloc[0]
print(f'Embarked mode: {embarked_mode}')
df_drop['embarked'] = df_drop['embarked'].fillna(embarked_mode)

# Fare — fill with mode
fare_mode = df_drop['fare'].mode().iloc[0]
print(f'Fare mode: {fare_mode}')
df_drop['fare'] = df_drop['fare'].fillna(fare_mode)
````
